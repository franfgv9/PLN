🧠 1️⃣ ¿Qué es un word embedding?

Un word embedding es una representación matemática (vectorial) de una palabra.
Cada palabra se transforma en un vector de números — por ejemplo, 100 números reales (de ahí “100 dimensiones”).

La idea es que palabras con significados parecidos tengan vectores parecidos.

📊 Ejemplo conceptual:


Palabra	Vector simplificado (3D)

gato	[0.12, -0.45, 0.78]

cão	[0.10, -0.40, 0.79]

carro	[-0.60, 0.22, -0.11]

¿Qué es GloVe?

GloVe (Global Vectors for Word Representation) es un modelo de embeddings creado por Stanford.
Se basa en analizar grandes corpus de texto (como Wikipedia o textos de internet) para aprender automáticamente cómo las palabras aparecen juntas (coocurrencia).
De esa forma, aprende relaciones semánticas entre palabras.

🔹 Ejemplos de relaciones que el modelo puede aprender:

“rei – homem + mulher ≈ rainha”

“Paris – França + Itália ≈ Roma”

“andar” ≈ “caminhar”

“feliz” ≈ “alegre”

El modelo que usamos aquí fue entrenado por el NILC (Brasil) en portugués, por lo tanto entiende relaciones de palabras portuguesas.

¿Qué es Gensim?

Gensim es una biblioteca Python diseñada para:

Trabajar con word embeddings (como Word2Vec o GloVe).

Cargar, consultar y comparar vectores de palabras.

Calcular similaridades o analogías semánticas.

Es una herramienta práctica para aplicar modelos de lenguaje preentrenados, sin tener que entrenarlos desde cero.

¿Por qué esto es importante en PLN?

Porque los word embeddings son la base de la semántica computacional moderna.
Antes de modelos grandes como BERT o GPT, los embeddings eran la forma principal de que un ordenador “entendiera” palabras y relaciones entre ellas.

Incluso hoy, siguen siendo muy usados porque:

Permiten reducir la dimensionalidad del texto.

Capturan significado y contexto de forma matemática.

Se usan en tareas como clasificación de sentimientos, traducción, recomendación de palabras, etc.

🧩 Exercício 1.1 — Similaridade entre pares de palavras

🎯 Objetivo

Calcular la similaridad semántica entre todas las combinaciones posibles de un conjunto de palabras:

🧠 Explicación

La similaridad entre dos palabras mide qué tan relacionadas están sus representaciones vectoriales.
En Gensim, se usa el coseno del ángulo entre los vectores.
El resultado va de -1 a 1:

1 → significados casi idénticos

0 → sin relación

-1 → significados opuestos

```python
# Lista de palavras
palavras = ['cão', 'gato', 'osso', 'carro', 'moto']

for p1 in palavras:
    for p2 in palavras:
        if p1 != p2:
            similaridade = model.similarity(p1, p2)
            print(f"Similaridade entre {p1} e {p2}: {similaridade}")
```

Exercício 1.4 — Funções com analogias

🎯 Objetivo

Crear funciones personalizadas que utilicen el principio de analogías semánticas (el mismo que acabas de aprender)
para encontrar palabras relacionadas según distintos tipos de relación lingüística, por ejemplo:

Masculino → Feminino

Normal → Diminutivo

Infinitivo → Gerúndio

Singular → Coletivo

País → Capital

🧠 1️⃣ Cómo lo haremos

Cada una de estas relaciones puede representarse como una analogía vectorial, igual que “rei - homem + mulher ≈ rainha”.
Por ejemplo:

homem → mulher define una dirección de género.

livro → livrinho define una dirección de diminutivo.

andar → andando define una dirección verbal (infinitivo → gerúndio).

La idea es:

Definir un par de referência (por ejemplo, homem–mulher).

Usar ese par para aplicar la mesma transformação a outra palavra.

```python
# 🧩 Masculino → Feminino
def masc_fem(model, palavra):
    return model.most_similar(positive=[palavra, 'mulher'], negative=['homem'])

# 🧩 Normal → Diminutivo
def norm_dim(model, palavra):
    return model.most_similar(positive=[palavra, 'pequenino'], negative=['pequeno'])

# 🧩 Infinitivo → Gerúndio
def inf_ger(model, palavra):
    return model.most_similar(positive=[palavra, 'comendo'], negative=['comer'])

# 🧩 Singular → Coletivo
def sing_col(model, palavra):
    return model.most_similar(positive=[palavra, 'manada'], negative=['vaca'])

# 🧩 País → Capital
def pais_capital(model, palavra):
    return model.most_similar(positive=[palavra, 'lisboa'], negative=['portugal'])

# 🧠 Lista de palavras para testar
palavras = ['cavalo', 'bom', 'estudar', 'cão', 'franca']

# 🔍 Testes automáticos
for p in palavras:
    print(f"Feminino de {p}: {masc_fem(model, p)}\n")
    print(f"Diminutivo de {p}: {norm_dim(model, p)}\n")
    print(f"Gerúndio de {p}: {inf_ger(model, p)}\n")
    print(f"Coletivo de {p}: {sing_col(model, p)}\n")
    print(f"Capital de {p}: {pais_capital(model, p)}\n")
    print("—" * 60)
```

Exercício 1.5 — Sentimento de Palavras

🎯 Objetivo

Criar uma função que indique se uma palavra transmite um sentimento positivo ou negativo,
com base na sua similaridade com palavras que carregam emoções típicas.

```python
# OPCIÓN CHATGPT

# === EXERCÍCIO 1.5: Sentimento de Palavras ===
from gensim.models import KeyedVectors

# Carregar o modelo
model = KeyedVectors.load_word2vec_format('glove_s100.txt')
print("Modelo carregado com sucesso!\n")

# 🧩 Função para determinar o sentimento
def sentimento_palavra(palavra):
    try:
        # Palavras de referência
        positivas = ['bom', 'feliz', 'alegre', 'ótimo', 'positivo']
        negativas = ['mau', 'triste', 'horrível', 'terrível', 'negativo']

        # Média da similaridade com palavras positivas e negativas
        sim_pos = sum(model.similarity(palavra, p) for p in positivas) / len(positivas)
        sim_neg = sum(model.similarity(palavra, n) for n in negativas) / len(negativas)

        # Determinar o resultado
        if sim_pos > sim_neg:
            sentimento = "Positivo 😀"
        elif sim_neg > sim_pos:
            sentimento = "Negativo 😞"
        else:
            sentimento = "Neutro 😐"

        return {
            'palavra': palavra,
            'similaridade_positiva': round(sim_pos, 4),
            'similaridade_negativa': round(sim_neg, 4),
            'classificação': sentimento
        }
    except KeyError:
        return f"A palavra '{palavra}' não está no vocabulário do modelo."

# 🧠 Teste com algumas palavras
palavras_teste = ['feliz', 'triste', 'excelente', 'horrível', 'doente', 'forte', 'fraco']

for p in palavras_teste:
    resultado = sentimento_palavra(p)
    print(resultado)
```

```python
# OPCIÓN PROFESORA --> MÁS FÁCIL


# 🧩 Função de sentimento
def sentimento(modelo, palavra):
    pos = modelo.similarity('feliz', palavra)
    neg = modelo.similarity('triste', palavra)
    print("pos:", pos, "neg:", neg)
    return pos - neg  # positivo → sentimento positivo, negativo → sentimento negativo

# 🧠 Testes
print("comer:", sentimento(model, 'comer'), "\n")
print("dormir:", sentimento(model, 'dormir'), "\n")
print("correr:", sentimento(model, 'correr'), "\n")
print("cair:", sentimento(model, 'cair'), "\n")
print("dor:", sentimento(model, 'dor'), "\n")
print("natal:", sentimento(model, 'natal'), "\n")
```

Exercício 1.6 — Representação de Frases e Similaridade

🎯 Objetivo

Criar uma função que:

Represente uma frase através do vetor médio das suas palavras.

Calcule a similaridade entre frases com base nesses vetores médios.

Assim podemos verificar quão parecidas semanticamente são duas frases.

🧠 1️⃣ Conceito teórico

Cada palavra no modelo GloVe é representada por um vetor numérico (por exemplo, de 100 dimensões).
Para representar uma frase, podemos:

Obter o vetor de cada palavra da frase.

Calcular a média desses vetores.

vetor_frase = 1/n * sumatorio desde i=1 hasta n de (vector_palavra de i)
	​


Isso gera um único vetor que representa o “significado global” da frase.

Depois, com a função cosine_similarity da biblioteca scipy, podemos medir a proximidade entre duas frases.

```python
# OPCION CHATI

# === EXERCÍCIO 1.6: Representação e Similaridade de Frases ===
from gensim.models import KeyedVectors
from scipy.spatial.distance import cosine
import numpy as np

# Carregar o modelo
model = KeyedVectors.load_word2vec_format('glove_s100.txt')
print("Modelo carregado com sucesso!\n")

# 🧩 Função para representar uma frase pelo vetor médio
def vetor_frase(modelo, frase):
    palavras = frase.lower().split()
    vetores = []
    for p in palavras:
        if p in modelo.key_to_index:   # verifica se a palavra existe no vocabulário
            vetores.append(modelo[p])
    if len(vetores) == 0:
        return np.zeros(modelo.vector_size)
    return np.mean(vetores, axis=0)

# 🧩 Função para calcular similaridade entre duas frases
def similaridade_frases(modelo, frase1, frase2):
    v1 = vetor_frase(modelo, frase1)
    v2 = vetor_frase(modelo, frase2)
    # Similaridade do cosseno (quanto mais próxima de 1, mais parecidas)
    sim = 1 - cosine(v1, v2)
    return sim

# 🧠 Testes
frases = [
    ("O gato dorme no sofá", "O cachorro dorme na cama"),
    ("Estou muito feliz hoje", "Estou triste e cansado"),
    ("O carro é rápido", "O automóvel é veloz"),
    ("Ela gosta de música", "Ele odeia barulho")
]

for f1, f2 in frases:
    print(f"\nFrase 1: {f1}")
    print(f"Frase 2: {f2}")
    print(f"Similaridade: {similaridade_frases(model, f1, f2):.4f}")
```
