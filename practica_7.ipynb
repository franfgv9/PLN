🤖 2. ¿Qué es Ollama y por qué se usa en esta práctica?

Ollama es un servidor local para modelos LLM.
La práctica lo utiliza porque permite:

- Ejecutar modelos sin depender de OpenAI u otras APIs.

- Controlar parámetros (temperature, modelo, etc.).

- Integrarlo fácilmente con LangChain, que es la librería que usarás para hacer prompting en Python.

Esta práctica quiere que entiendas cómo funcionan los LLM a bajo nivel, no sólo a través de ChatGPT.
Por eso se usa Ollama: te obliga a interactuar directamente con el modelo, gestionando instalación, servidor y parámetros.

✅ 1. Si ya tienes Ollama instalado localmente en Windows, ¿también necesitas instalar LangChain?

Sí.
La instalación de LangChain-Ollama ES NECESARIA incluso si estás trabajando de forma local.

¿Por qué?
Porque Ollama por sí solo solo ejecuta modelos.
Pero la práctica usa Python, y específicamente LangChain, para:

- invocar modelos desde código

- controlar la temperatura

- estructurar prompts

- ejecutar ejemplos de few-shot, system prompts, chain-of-thought, etc.
